```python
print(
    "전세인데 월세가 0이 아닌 건수:",
    (
        (df_apt_2023["전월세구분"] == "전세")
        & (df_apt_2023["임대료(만원)"] != 0)
    ).sum()
)

print(
    "월세인데 월세가 0인 건수:",
    (
        (df_apt_2023["전월세구분"] == "월세")
        & (df_apt_2023["임대료(만원)"] == 0)
    ).sum()
)
```
